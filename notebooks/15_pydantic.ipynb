```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name : str
    age : int
```

```python
user = User(name = "Pooja", age=23)
user
```

## Field validation

```python
class Product(BaseModel):
    title : str = Field(max_length=10)
    des : str = Field(min_length=20)
    price : int = Field(gt=0, le=100)
```

```python
product = Product(title ="iphone 15", des = "this is amazing mobile phone from apple...", price = 20)
product2 = Product(title ="iphone 17", des = "this is amazing mobile phone from apple...", price = 40)

product
```

```python
class Order(BaseModel):
    id : str = Field(min_length=6, max_length=12)
    product : Product
```

```python
order = Order(id = "a1b2c3", product=product)
order
```

```python
order.id
```

```python
order.product.des
```

```python
class Cart(BaseModel):
    id : str = Field(min_length=6, max_length=12)
    product : List[Product]
    comment : str = Field(default = "no comment")

cart = Cart(id = "12abcdef", product = [product, product2])
print(cart)
```

```python
for product in cart.product:
    print(product.title)
```

```python
from typing import Annotated
from pydantic import BaseModel, Field

# 1. Define your reusable domain types
Username = Annotated[str, Field(min_length=3, max_length=20, pattern=r"^[a-zA-Z0-9_]+$")]
PositiveInt = Annotated[int, Field(gt=0)]

# 2. Use them cleanly inside your models
class User(BaseModel):
    username: Username
    age: PositiveInt

class Product(BaseModel):
    sku: Username
    price: PositiveInt
```
