```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "Machine learning is a subset of artificial intelligence",
    "Deep learning uses neural networks with multiple layers",
    "Python is a programming language popular for data science",
    "India won the cricket world cup in 2011",
    "AI models can understand and generate human language"
]
```

```python
pip install -qU langchain-google-genai
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embeddings = GoogleGenerativeAIEmbeddings(model = "gemini-embedding-2-preview")
```

```python
vector = embeddings.embed_documents(documents)
len(vector[0])
```

```python
pip install -qU langchain-chroma
```

## Store vector embeddings in vector DB

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts = documents,
    embedding = embeddings,
    persist_directory="./vector_db"
)
```

```python
query = "Machine learning"
result = vector_store.similarity_search(query=query)
print(result)
```

```python
query = "Data Science"
result = vector_store.similarity_search(query=query, k = 3)
print(result)
```
