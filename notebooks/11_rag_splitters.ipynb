```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
texts = """Today, I want to speak from the heart about three pillars that define the future of our nation: our youth, our education system, and our farmers.
"""
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 50, chunk_overlap = 20)
```

```python
chunks = splitter.split_text(texts)
```

## Document text loaders

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
txt_loader = TextLoader("/Users/poojaveluri/Documents/genai/data/speech.txt")
text = txt_loader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)
```

```python
text_splitter = splitter.split_documents(text)
len(text_splitter)
```

```python
pdf_loader = PyPDFLoader("/Users/poojaveluri/Documents/genai/data/medical_report.pdf")
pdf = pdf_loader.load()
len(pdf)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)
```

```python
pdf_chunks = splitter.split_documents(pdf)
len(pdf_chunks)
```
