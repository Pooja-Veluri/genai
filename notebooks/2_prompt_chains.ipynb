### Complete code of prompt template in langchain

```python
from langchain_core.prompts import ChatPromptTemplate

prompts = ChatPromptTemplate.from_messages([
    ("system", "Translate the following text into {language}."),
    ("human", "{question}")
])

final_prompts = prompts.format_messages(
    language="French",
    question="I love visiting new places and trying different cuisines."
)

#print(final_prompts)

response = llm.invoke(final_prompts)
print(response.text)
```

### Chains

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompts = ChatPromptTemplate.from_messages([
    ("system", "Translate the input text into {language}."),
    ("human", "{question}")
])
output_parser = StrOutputParser()

chains = prompts | llm | output_parser

response = chains.invoke({"language":"French", "question":"I love visiting new places and trying different cuisines."})

print(response)
```

### Transform case

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

def tranform_case(text: str) -> str:
    return text.lower()

prompts = ChatPromptTemplate.from_messages([
    ("system", "Translate the input text into {language}."),
    ("human", "{question}")
])
output_parser = StrOutputParser()

chains = prompts | llm | output_parser | tranform_case

response = chains.invoke({"language":"French", "question":"I love visiting new places and trying different cuisines."})

print(response)
```
