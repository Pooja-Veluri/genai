### Connect with Groq model

```python
%pip install -U langchain-groq python-dotenv
```

```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

response = llm.invoke("What is RAG? explain clearly in simple terms not more than 100 words")
print(response.content)
```

### Static Prompt


```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
prompts =[
    ("system: You're a helpful python programming assistant"),
    ("user: How to sort a list in python?"),
]

response = llm.invoke(prompts)
print(response.text)
```
